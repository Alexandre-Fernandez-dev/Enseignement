```python
from sympy import *
from sympy import init_printing
from sympy import re, im, I, E, conjugate, symbols
from spb import plot, BB
init_printing()
```

```python
x = symbols("x")
l = interpolate([(1, -1), (2, 1), (0, 1)], x)
l
```

```python
d = diff(l)
d
```

```python
def tangent(e, a):
    d = diff(e)
    return d.subs(x, a)*(x - a) + e.subs(x, a)
```

```python
t_2 = tangent(l, 2)
#t_2 = tangent(l, 2)
```

```python
p1 = plot(l, t_2, xlim=(0,3), ylim=(-1.5,2), label=["$C_f$", "$T_2$"], xlabel=" ", ylabel=" ", rendering_kw=[{"linestyle": "-"}], adaptive=False, is_point=False,axis_center="auto")
```

```python
p1.save("functionA.png", dpi=200)
```
