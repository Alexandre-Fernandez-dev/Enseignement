```python
from sympy import *
from sympy import init_printing
from sympy import re, im, I, E, conjugate, symbols
from spb import plot, BB
init_printing()
```

```python
x = symbols("x")
f = -4*(x-1)*(x-2)
expand(f)
```

```python
p = plot(f, xlim=(0,3), ylim=(-1,1.5), label=["f"], xlabel=" ", ylabel=" ", rendering_kw=[{"linestyle": "-"}], adaptive=False, is_point=False,axis_center="auto")
```

```python
p.save("functionA.png", dpi=200)
```

```python
g = 2*(x-1)*(x-3)
expand(g)
```

```python
p2 = plot(g, xlim=(0,4), ylim=(-2.5,1), label=["f"], xlabel=" ", ylabel=" ", rendering_kw=[{"linestyle": "-"}], adaptive=False, is_point=False,axis_center="auto")
```

```python
p2.save("functionB.png", dpi=200)
```
