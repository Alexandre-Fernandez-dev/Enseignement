```python
from sympy import *
from sympy import init_printing
from sympy import re, im, I, E, conjugate, symbols
from spb import plot, BB
init_printing()
```

```python
x = symbols("x")
l = interpolate([(1, -1), (2, 0.5), (0, 0.5)], x)
l
```

```python
d = diff(l)
d
```

```python
def tangent(e, a):
    d = diff(e)
    return d.subs(x, a)*(x - a) + e.subs(x, a)
```

```python
t_1 = tangent(l, 1)
t_2 = tangent(l, 2)
```

```python
p1 = plot(l, t_1, t_2, xlim=(0,3), ylim=(-1.5,2), label=["$C_f$", "$T_1$", "$T_2$"], xlabel=" ", ylabel=" ", rendering_kw=[{"linestyle": "-"}], adaptive=False, is_point=False,axis_center="auto")
```

```python
p1.save("functionA.png", dpi=200)
```
